# เก็บข้อมูลดิบของ Test subjects (ยังไม่ผ่าน preprocess)
คัดลอกไฟล์ HMC ต้นฉบับของ Test 23 subjects ไปไว้ที่ `dataset/Test_Full_night` เพื่อให้ไฟล์อื่นนำไปสร้าง Hypnogram เทียบเส้นจริงทั้งคืน (Ground Truth) กับค่าที่โมเดลทำนาย

```python
from pathlib import Path
import shutil

PROJECT_DIR = "/content/drive/MyDrive/SleepStageClassificationProject"
DATA_DIR = Path(f"{PROJECT_DIR}/dataset/HMC")
OUT_DIR = Path(f"{PROJECT_DIR}/dataset/Test_Full_night")
OUT_DIR.mkdir(parents=True, exist_ok=True)

test_subject_list = [
    "SN010", "SN013", "SN020", "SN021", "SN024", "SN028", "SN031", "SN038",
    "SN047", "SN058", "SN067", "SN079", "SN084", "SN085", "SN087", "SN101",
    "SN114", "SN117", "SN136", "SN137", "SN141", "SN146", "SN147",
]
print("Test subjects:", len(test_subject_list))
```

## คัดลอกไฟล์ดิบของแต่ละ subject
คัดลอกทุกไฟล์ที่ขึ้นต้นด้วยชื่อ subject (เช่น สัญญาณ `.edf` และไฟล์ sleep scoring) โดยไม่แก้ไขเนื้อหา ถ้าไฟล์มีอยู่แล้วจะข้ามไป จึงรันซ้ำได้

```python
missing = []
for subject in test_subject_list:
    files = sorted(p for p in DATA_DIR.rglob(f"{subject}*") if p.is_file())
    if not files:
        missing.append(subject)
        print(f"{subject}: ไม่พบไฟล์")
        continue
    for src in files:
        dst = OUT_DIR / src.name
        if dst.exists() and dst.stat().st_size == src.stat().st_size:
            continue
        shutil.copy2(src, dst)
    print(f"{subject}: {[p.name for p in files]}")

print("\nไม่พบไฟล์:", missing if missing else "ไม่มี")
```

## ตรวจสอบผลลัพธ์

```python
saved = sorted(OUT_DIR.glob("*"))
n_subject = len({p.name.split(".")[0].split("_")[0] for p in saved})
total_gb = sum(p.stat().st_size for p in saved) / 1e9
print(f"ใน {OUT_DIR.name}: {len(saved)} ไฟล์ จาก {n_subject} subjects | ขนาดรวม {total_gb:.2f} GB")
for p in saved[:69]:
    print(" ", p.name)
```
